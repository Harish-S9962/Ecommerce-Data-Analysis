```python
#which category generates the most revenue.

revenue_by_category = orders.groupby('category')['net_amount'].sum().sort_values(ascending=False)

print(revenue_by_category)
```

```python
import matplotlib.pyplot as plt

revenue_by_category.plot(kind='bar')

plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue')

plt.xticks(rotation=45)

plt.show()
```

```python
# Which category has the highest number of orders?

orders_by_category = orders.groupby('category')['order_id'].nunique().sort_values(ascending=False)

print(orders_by_category)
```

```python
orders_by_category.plot(kind='bar')

plt.title('Orders by Category')
plt.xlabel('Category')
plt.ylabel('Number of Orders')

plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# ORDER BY STATE
orders_with_customer = orders.merge(
    customers[['customer_id','customer_segment', 'state']],
    on='customer_id',
    how='left'
)
```

```python
orders_by_state = orders_with_customer.groupby('state')['order_id'].nunique().sort_values(ascending=False)

print(orders_by_state)
```

```python
# REVENUE BY STATE
Revenue_by_state = orders_with_customer.groupby('state')['net_amount'].sum().sort_values(ascending=False).round(0)

print(orders_by_state)
```

```python
Revenue_by_state.plot(kind='bar')

plt.title('Revenue by State')
plt.xlabel('State')
plt.ylabel('Revenue')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Orders by City

orders_by_city = orders.groupby('delivery_city')['order_id'].nunique().sort_values(ascending=False)

print(orders_by_city)
```

```python
top_10_cities = orders_by_city.head(10)

top_10_cities.plot(kind='bar')

plt.title('Top 10 Cities by Orders')
plt.xlabel('City')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# REVENUE BY CITY
revenue_by_city = orders.groupby('delivery_city')['net_amount'].sum().sort_values(ascending=False)

print(revenue_by_city)
```

```python
top_10_city_revenue = revenue_by_city.head(10)

top_10_city_revenue.plot(kind='bar')

plt.title('Top 10 Cities by Revenue')
plt.xlabel('City')
plt.ylabel('Revenue')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
# ORDER AND REVENUE BY CUSTOMER SEGMENT

segment_analysis = orders_with_customer.groupby('customer_segment').agg(
    Orders=('order_id', 'nunique'),
    Revenue=('net_amount', 'sum')
).sort_values('Revenue', ascending=False).round(0)

print(segment_analysis)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

# Orders
ax1.bar(
    segment_analysis.index,
    segment_analysis['Orders'],
    alpha=0.7,
    label='Orders'
)

ax1.set_xlabel('Customer Segment')
ax1.set_ylabel('Number of Orders')
ax1.tick_params(axis='x', rotation=45)

# Revenue
ax2 = ax1.twinx()

ax2.plot(
    segment_analysis.index,
    segment_analysis['Revenue'],
    marker='o',
    linewidth=2,
    label='Revenue'
)

ax2.set_ylabel('Revenue')

plt.title('Orders and Revenue by Customer Segment')

plt.tight_layout()
plt.show()
```

```python
# CATEGORY DISCOUNT

discount_analysis = orders.groupby('category').agg(
    Orders=('order_id', 'nunique'),
    Revenue=('net_amount', 'sum'),
    Discount=('discount_amount', 'sum')
).sort_values('Discount', ascending=False).round(0)

print(discount_analysis)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

# Revenue - Bar
ax1.bar(
    discount_analysis.index,
    discount_analysis['Revenue'],
    alpha=0.7
)

ax1.set_xlabel('Category')
ax1.set_ylabel('Revenue')
ax1.tick_params(axis='x', rotation=45)

# Discount - Line
ax2 = ax1.twinx()

ax2.plot(
    discount_analysis.index,
    discount_analysis['Discount'],
    marker='o',
    linewidth=2
)

ax2.set_ylabel('Discount Amount')

plt.title('Revenue and Discount by Category')

plt.tight_layout()
plt.show()
```

```python
#ORDER STATUS ANALYSIS

status_analysis = orders.groupby('order_status').agg(
    Orders=('order_id', 'nunique'),
    Revenue=('net_amount', 'sum')
).sort_values('Orders', ascending=False).round(0)

print(status_analysis)
```

```python
status_analysis['Order_Percentage'] = (
    status_analysis['Orders'] / status_analysis['Orders'].sum() * 100
)

print(status_analysis)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

# Orders - Bar
ax1.bar(
    status_analysis.index,
    status_analysis['Orders'],
    alpha=0.7
)

ax1.set_xlabel('Order Status')
ax1.set_ylabel('Number of Orders')

# Revenue - Line
ax2 = ax1.twinx()

ax2.plot(
    status_analysis.index,
    status_analysis['Revenue'],
    marker='o',
    linewidth=2
)

ax2.set_ylabel('Revenue')

plt.title('Orders and Revenue by Order Status')

plt.tight_layout()
plt.show()
```

```python
# AGE WISE ORDERS

customers['age_group'] = pd.cut(
    customers['age'],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=['18-25', '26-35', '36-45', '46-55', '56-65', '66+']
)
```

```python
orders_with_customer = orders.merge(
    customers[['customer_id', 'age_group']],
    on='customer_id',
    how='left'
)
```

```python
age_analysis = orders_with_customer.groupby(
    'age_group',
    observed=False
).agg(
    Orders=('order_id', 'nunique'),
    Revenue=('net_amount', 'sum')
).sort_values('Orders', ascending=False).round(0)


print(age_analysis)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

# Orders - Bar
ax1.bar(
    age_analysis.index.astype(str),
    age_analysis['Orders'],
    alpha=0.7
)

ax1.set_xlabel('Age Group')
ax1.set_ylabel('Number of Orders')

# Revenue - Line
ax2 = ax1.twinx()

ax2.plot(
    age_analysis.index.astype(str),
    age_analysis['Revenue'],
    marker='o',
    linewidth=2
)

ax2.set_ylabel('Revenue')

plt.title('Orders and Revenue by Age Group')

plt.tight_layout()
plt.show()
```

```python
# YAER BY REVENUE AND ORDERS

orders['order_year'] = orders['order_date'].dt.year

print(orders['order_year'].unique())
```

```python
year_analysis = orders.groupby('order_year').agg(
    Orders=('order_id', 'nunique'),
    Revenue=('net_amount', 'sum')
).sort_values('Revenue',ascending=False).round(0)

print(year_analysis)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

# Orders - Bar
ax1.bar(
    year_analysis.index.astype(str),
    year_analysis['Orders'],
    alpha=0.7
)

ax1.set_xlabel('Year')
ax1.set_ylabel('Number of Orders')

# Revenue - Line
ax2 = ax1.twinx()

ax2.plot(
    year_analysis.index.astype(str),
    year_analysis['Revenue'],
    marker='o',
    linewidth=2
)

ax2.set_ylabel('Revenue')

plt.title('Year-wise Orders and Revenue')

plt.tight_layout()
plt.show()
```

```python
# REVENUE VS QUANTITY

revenue_quantity = orders.groupby('category').agg(
    Quantity=('quantity', 'sum'),
    Revenue=('net_amount', 'sum')
).sort_values('Revenue',ascending=False).round(0)

print(revenue_quantity)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    revenue_quantity['Quantity'],
    revenue_quantity['Revenue']
)

plt.xlabel('Quantity Sold')
plt.ylabel('Revenue')
plt.title('Revenue vs Quantity by Category')

plt.tight_layout()
plt.show()
```

```python
# GENDER ANALYSIS

gender_analysis = orders.merge(
    customers[['customer_id', 'gender']],
    on='customer_id',
    how='left'
).groupby('gender').agg(
    Orders=('order_id', 'nunique'),
    Revenue=('net_amount', 'sum')
).sort_values('Revenue', ascending=False).round(0)

print(gender_analysis)
```

```python
fig, ax1 = plt.subplots(figsize=(10, 6))

ax1.bar(
    gender_analysis.index,
    gender_analysis['Orders'],
    alpha=0.7
)

ax1.set_xlabel('Gender')
ax1.set_ylabel('Number of Orders')

ax2 = ax1.twinx()

ax2.plot(
    gender_analysis.index,
    gender_analysis['Revenue'],
    marker='o',
    linewidth=2
)

ax2.set_ylabel('Revenue')

plt.title('Orders and Revenue by Gender')

plt.tight_layout()
plt.show()
```
